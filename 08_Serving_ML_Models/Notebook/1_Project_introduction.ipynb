## Topic: Serving the ML Model 

### Introduction of Project Workflow

- 1. Model Building
    - save the model with name model.pkl

- 2. Create a FastAPI 
    - Create a endpoint that pass the input to model and get the output of the model.
    


- 3. Create a Frontend(UI)
    - Using the Streamlit that is connect to the FastAPI.
    - take the input from the user and pass it to the FastAPI
    - Get the model output via FastAPI and Display the output to the user.


- Key Rule of Thump:
    - model.pkl <---> FastAPI <----> Fronted


#### 2. Create a FastAPI 
- Here, the HTTP method is Post.

- Step 1: Create an Endpoint 
    - /predict
    - request body:
        - input: an individual person all facts or information.
        - eg: age, weight, height, income_lpa, smoker, city, occupation (raw data)

- Step 2: Validate the user input and create a new features that our model take
    - Validate using pydantic class
    
    - create a new features: computed_field 


- Step 3: Pass the Validated and create new feature to the Model
    - provide the Json response based on Provided inputs
