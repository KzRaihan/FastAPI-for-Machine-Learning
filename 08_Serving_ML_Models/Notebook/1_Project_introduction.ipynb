## Topic: Serving the ML Model 

### Introduction of Project Workflow

- 1. Model Building
    - save the model with name model.pkl

- 2. Create a FastAPI 
    - Create a endpoint that pass the input to model and get the output of the model.


- 3. Create a Frontend(UI)
    - Using the Streamlit that is connect to the FastAPI.
    - take the input from the user and pass it to the FastAPI
    - Get the model output via FastAPI and Display the output to the user.


- Key Rule of Thump:
    - model.pkl <---> FastAPI <----> Fronted
