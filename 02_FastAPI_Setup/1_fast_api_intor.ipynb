### 1. Introduction of FastApI
- Definition:
    - FastAPI is a model python framework use for building high-Performance APIs with automatic validation and documentation.

- FastAPI build using python two libraries
    - 1. Starlette
    - 2. Pydantic

- 1. Rule of Starlette in FastAPI
    - Starlette manages how our API receives requests and sends back responses.

    - Starlette is a lightweight, high-performance ASGI (Asynchronous Server Gateway Interface) framework and toolkit for building asynchronous Python web services.

    - Starlette serves as the fundamental HTTP engine and toolkit that powers FastAPI.


- 2. Rule of Pydantic in FastAPI
    - It is data validation library.
    - Pydantic is used to check if the data coming into our API is correct and in the right format.

```python
""" -> The 5 Pillars of FastAPI:

┌──────────────────────────────────────────────────────────────┐
│                                                              │
│  1. FAST PERFORMANCE                                         │
│     Built on Starlette (ASGI) and Uvicorn                    │
│     One of the fastest Python frameworks                     │
│     On par with NodeJS and Go                                │
│                                                              │
│  2. FAST TO CODE                                             │
│     Reduce code duplication                                  │
│     Intuitive design                                         │
│     Fewer bugs                                               │
│                                                              │
│  3. AUTOMATIC DOCUMENTATION                                  │
│     Swagger UI at /docs                                      │
│     ReDoc at /redoc                                          │
│     OpenAPI schema auto-generated                            │
│                                                              │
│  4. TYPE SAFETY                                              │
│     Python type hints drive everything                       │
│     Editor support (autocomplete, error checking)            │
│     Runtime validation via Pydantic                          │
│                                                              │
│  5. STANDARDS-BASED                                          │
│     OpenAPI (formerly Swagger) specification                 │
│     JSON Schema                                              │
│     OAuth2 support                                           │
│                                                              │
└──────────────────────────────────────────────────────────────┘

"""
```

### 2. Architecture compare between Flask and FastAP

```python
""" 
- Architecture compare between Flask and FastAP
===============================================

- 1. Flask:
-----------

--------------                        --------------
| Web server |    -----> SGI --------> | API CODE |
-------------                         --------------

- Web server:
    - here, use the web server as Gunicorn.
    - Gunicorn is a WSGI HTTP server, specifically designed for python web application.
    


- SGI:
    - here SGI type is: WSGI [web Server Gateway Interface]
    - WSGI use the Werkzeug library.
    - Follow Synchronous Communication approach (one after another ).
    - so that it's slower request processing and scalability challenges.

- API CODE:
    - Python written code
    - 

- Problem of Flask:
    - 1. Performance Issues.
    - 2. Waiting for I/O times and high latency.
    - 3. Can arise misconfiguration.
    - 4. Debugging can be challenging.

"""
```

```python
""" 

- 2. FastAPI:
-----------

--------------                        --------------
| Web server |    -----> SGI --------> | API CODE  |
-------------                         --------------

- Web server:
    - here, use the web server as unicorn.
    - unicorn is a ASGI HTTP server, specifically designed for python web application.
    - It is a high-performance ASGI server.
    


- SGI:
    - here SGI type is: ASGI [Asynchronous Server Gateway Interface]
    - WSGI use the Starlette library.
    - Follow Asynchronous Communication approach (at a same time process multiple request).
    - so that it's fast request processing.

- API CODE:
    - Python written code
    

- Benefit of FastAPI:
    - 1. FAST PERFORMANCE
    - 2. AUTOMATIC DOCUMENTATION  
    - 3. TYPE SAFETY 
    - 4. Debugging is Easy.
    - 5. Seamless Integration with Modern Ecosystem (ML/DL libraries, OAuth, JWT, SQL Alchemy, Docker, Kubernetes etc)

"""
```

```python
""" 
┌──────────────────────────────────────────────────────────────┐
│                                                              │
│  fastapi          → The web framework                        │
│                     (routing, validation, docs)              │
│                                                              │
│  uvicorn          → The ASGI server                          │
│                     (actually runs your app,                 │
│                      handles HTTP connections)               │
│                                                              │
│  [standard]       → Extra Uvicorn features                   │
│                     (auto-reload, better performance,        │
│                      WebSocket support)                      │
│                                                              │
│  Analogy:                                                    │
│    FastAPI = The car (engine, steering, features)            │
│    Uvicorn = The driver (actually drives the car)            │
│                                                              │
└──────────────────────────────────────────────────────────────┘

"""
```

```python
# 1. import library
from fastapi import FastAPI
```

```python
# 2. create an web application object

app = FastAPI()
```
