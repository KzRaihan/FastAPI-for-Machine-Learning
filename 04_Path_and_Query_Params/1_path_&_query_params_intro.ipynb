### 1. Introduction of Path Params
- Definition:
    - Path parameters are dynamic segments of a URL path used to identify a specific resource.


    - eg: 
        - localhost:8080/view/3
        - here, 
            - 3 is the path params that can be changes on dynamic
            - It is very useful when we retrieve, update and delete the specific data.
            
    
- Key Rule Thump:
    - Path parameters are variable parts of the URL path that capture values to identify specific resources.

```python
""" 
- FastAPI uses Python type hints to automatically validate path parameters: 

┌──────────────────────────────────────────────────────────────┐
│          AUTOMATIC TYPE VALIDATION IN ACTION                 │
│                                                              │
│  @app.get("/items/{item_id}")                                │
│  def get_item(item_id: int):                                 │
│                                                              │
│  Request               Result                                │
│  ─────────────────────────────────────────────────────       │
│  GET /items/42            item_id = 42 (int)                 │
│  GET /items/0             item_id = 0 (int)                  │
│  GET /items/-5            item_id = -5 (int)                 │
│  GET /items/3.14          422 Error (not an integer)         │
│  GET /items/abc           422 Error (not an integer)         │
│  GET /items/              404 Error (no path match)          │
│                                                              │
│  You get validation FOR FREE. No manual checks needed!       │
│                                                              │
└──────────────────────────────────────────────────────────────┘

"""
```

### 2. Introduction of Query Params

- Definition:
    - Query parameters appear after the ? in a URL and are used for filtering, sorting, and pagination.

```python
""" 
┌──────────────────────────────────────────────────────────────┐
│          REQUIRED vs OPTIONAL CHEAT SHEET                    │
│                                                              │
│  Syntax                          Required?  Default          │
│  ─────────────────────────────────────────────────────       │
│  param: str                      YES        None             │
│  param: int                      YES        None             │
│  param: Optional[str] = None     NO         None             │
│  param: str = "default"          NO         "default"        │
│  param: int = 10                 NO         10               │
│  param: str = Query(...)         YES        None             │
│  param: str = Query("default")   NO         "default"        │
│                                                              │
│  RULE: If there's NO default value → it's REQUIRED           │
│        If there IS a default value → it's OPTIONAL           │
│                                                              │
└──────────────────────────────────────────────────────────────┘

"""
```

```python
""" 
┌──────────────────────────────────────────────────────────────────┐
│                  Query() PARAMETERS                              │
│                                                                  │
│  Query(                                                          │
│      default,         # Default value or ... for required        │
│                                                                  │
│      # ── Numeric Constraints ──                                 │
│      gt=0,            # Greater Than                             │
│      ge=0,            # Greater than or Equal                    │
│      lt=100,          # Less Than                                │
│      le=100,          # Less than or Equal                       │
│      multiple_of=5,   # Must be multiple of this                 │
│                                                                  │
│      # ── String Constraints ──                                  │
│      min_length=1,    # Minimum length                           │
│      max_length=100,  # Maximum length                           │
│      pattern="^...",  # Regex pattern                            │
│                                                                  │
│      # ── Metadata ──                                            │
│      alias="name",    # URL parameter name (if different)        │
│      title="...",     # Short title for docs                     │
│      description="...", # Description for docs                   │
│      deprecated=True, # Mark as deprecated in docs               │
│                                                                  │
│      # ── List Behavior ──                                       │
│      # (see List Query Parameters section below)                 │
│  )                                                               │
│                                                                  │
└──────────────────────────────────────────────────────────────────┘

"""
```

```python
"""
- Path and Query Parameters:
============================

https://api.example.com:8000/models/42/metrics?status=active&limit=10
  │        │              │    │   │    │       │     │        │
  │        │              │    │   │    │       │     │        └── Query value
  │        │              │    │   │    │       │     └── Query key
  │        │              │    │   │    │       └── Query separator (&)
  │        │              │    │   │    └── Path parameter
  │        │              │    │   └── Nested path
  │        │              │    └── Path parameter
  │        │              └── Resource path
  │        └── Host
  └── Protocol

"""


""" - The Three Ways to Send Data to an API 


┌──────────────────────────────────────────────────────────────────┐
│            THREE WAYS TO SEND DATA IN HTTP                       │
│                                                                  │
│  1. PATH PARAMETERS (in the URL path)                            │
│     ┌──────────────────────────────────────────────────────┐     │
│     │  GET /models/42/metrics                              │     │
│     │              ↑↑                                      │     │
│     │         Path parameter                               │     │
│     │         Identifies a SPECIFIC resource               │     │
│     └──────────────────────────────────────────────────────┘     │
│                                                                  │
│  2. QUERY PARAMETERS (after the ? in the URL)                    │
│     ┌──────────────────────────────────────────────────────┐     │
│     │  GET /models?status=active&limit=10                  │     │
│     │                  ↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑                    │     │
│     │              Query parameters                        │     │
│     │              FILTER, SORT, PAGINATE                  │     │
│     └──────────────────────────────────────────────────────┘     │
│                                                                  │
│  3. REQUEST BODY (in the HTTP body — covered in Module 05)       │
│     ┌──────────────────────────────────────────────────────┐     │
│     │  POST /predict                                       │     │
│     │  Body: {"sepal_length": 5.1, "sepal_width": 3.5}     │     │
│     │         ↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑↑       │     │
│     │              Request body                            │     │
│     │              Complex data for CREATE/ACTION          │     │
│     └──────────────────────────────────────────────────────┘     │
│                                                                  │
└──────────────────────────────────────────────────────────────────┘

"""
```

```python
""" 
- When to Use What?
====================

┌──────────────────────┬──────────────────────────────────────────┐
│  Parameter Type      │  Use When...                             │
├──────────────────────┼──────────────────────────────────────────┤
│  Path Parameter      │  Identifying a SPECIFIC resource         │
│                      │  "Which ONE thing do you want?"          │
│                      │  /users/42  /models/iris  /files/abc     │
│                      │                                          │
│                      │  It's part of the resource IDENTITY      │
├──────────────────────┼──────────────────────────────────────────┤
│  Query Parameter     │  Filtering, sorting, paginating          │
│                      │  "How do you want to refine the results?"│
│                      │  ?status=active  ?page=2  ?sort=name     │
│                      │                                          │
│                      │  It's OPTIONAL refinement                │
├──────────────────────┼──────────────────────────────────────────┤
│  Request Body        │  Sending complex data                    │
│                      │  "Here's the data to process"            │
│                      │  POST /predict  {"features": {...}}      │
│                      │                                          │
│                      │  It's the PAYLOAD                        │
└──────────────────────┴──────────────────────────────────────────┘

"""
```

### 3. Summary

```python
""" 
┌──────────────────────────────────────────────────────────────────┐
│                  MODULE 04 — KEY TAKEAWAYS                       │
│                                                                  │
│  PATH PARAMETERS:                                                │
│  -  Identify specific resources: /models/{model_id}              │
│  -  Type hints auto-validate: int, str, float, bool              │
│  -  Path() adds constraints: gt, ge, lt, le, min_length, pattern │
│  -  Enums restrict to fixed values                               │
│  -  Multiple path params: /users/{uid}/models/{mid}              │
│  -  Route order: specific BEFORE parameterized                   │
│  -  Max 2-3 path params per route                                │
│                                                                  │
│  QUERY PARAMETERS:                                               │
│  -  Filter, sort, paginate: ?status=active&limit=10              │
│  -  No default = REQUIRED, with default = OPTIONAL               │
│  -  Optional[str] = None for truly optional params               │
│  -  Query() adds constraints and documentation                   │
│  -  List params: ?tags=a&tags=b → ["a", "b"]                     │
│  -  Aliases: Query(..., alias="q") → URL uses ?q=...             │
│  -  Auto-parsing: dates, booleans, enums                         │
│                                                                  │
│  COMBINED:                                                       │
│  -  Path = IDENTITY, Query = REFINEMENT                          │
│  -  FastAPI auto-detects based on URL template                   │
│  -  Use together for powerful, flexible APIs                     │
│                                                                  │
│  ML PATTERNS:                                                    │
│  -  /models/{name}                   → Model lookup              │
│  -  /models?task=...&framework=...   → Model search              │
│  -  /models/{name}/versions/{ver}    → Versioning                │
│  -  /monitoring/{name}/drift?type=.. → Monitoring                │
│  -  /predictions?start=...&end=...   → History                   │
│                                                                  │
└──────────────────────────────────────────────────────────────────┘

"""
```
