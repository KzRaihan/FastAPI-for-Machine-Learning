### 1. Introduction of HTTP Methods
- Definition of HTTP (HyperText Transfer Protocol):
    
    - HTTP is the protocol (set of rules) that governs how messages are formatted and transmitted between client and server on the web.



```python
"""
-> Client-Server Architecture:
==============================
┌─────────────────────────────────────────────────────────────────┐
│                   CLIENT-SERVER MODEL                           │
│                                                                 │
│   ┌────────────┐           Network             ┌────────────┐   │
│   │            │  ──── REQUEST ──────────────► │            │   │
│   │   CLIENT   │                               │   SERVER   │   │
│   │            │  ◄─── RESPONSE ─────────────  │            │   │
│   └────────────┘          (HTTP)               └────────────┘   │
│                                                                 │
│   Client:                          Server:                      │
│   • Initiates requests             • Listens for requests       │
│   • Displays data to user          • Processes requests         │
│   • Browser, Mobile App,           • Accesses database          │
│     Python script                  • Returns responses          │
│                                                                 │
└─────────────────────────────────────────────────────────────────┘



"""


""" 
┌──────────────────────────────────────────────────────┐
│                   HTTP REQUEST                       │
│                                                      │
│  ┌─────────────────────────────────────────────┐     │
│  │ REQUEST LINE                                │     │
│  │ POST /api/predict HTTP/1.1                  │     │
│  │  │      │            │                      │     │
│  │  │      │            └── HTTP Version       │     │
│  │  │      └─────────────── URL/Endpoint       │     │
│  │  └────────────────────── HTTP Method        │     │
│  └─────────────────────────────────────────────┘     │
│                                                      │
┌──────────────────────────────────────────────────────┐
│                   HTTP RESPONSE                      │
│                                                      │
│  ┌─────────────────────────────────────────────┐     │
│  │ STATUS LINE                                 │     │
│  │ HTTP/1.1 200 OK                             │     │
│  │    │       │   │                            │     │
│  │    │       │   └── Reason Phrase            │     │
│  │    │       └────── Status Code              │     │
│  │    └────────────── HTTP Version             │     │
│  └─────────────────────────────────────────────┘     │
└──────────────────────────────────────────────────────┘




"""
```

```python
""" 
┌──────────────────────────────────────────────────────────────────┐
│                    ALL HTTP METHODS                              │
│                                                                  │
│  ┌──────────────────────────────────────────────────────────┐    │
│  │  COMMON METHODS (You'll use these 99% of the time)       │    │
│  │                                                          │    │
│  │  GET     → Read / Retrieve                               │    │
│  │  POST    → Create / Submit                               │    │
│  │  PUT     → Replace / Full Update                         │    │
│  │  PATCH   → Modify / Partial Update                       │    │
│  │  DELETE  → Remove / Destroy                              │    │
│  └──────────────────────────────────────────────────────────┘    │
│                                                                  │
│  ┌──────────────────────────────────────────────────────────┐    │
│  │  LESS COMMON (Good to know)                              │    │
│  │                                                          │    │
│  │  HEAD    → Like GET but headers only (no body)           │    │
│  │  OPTIONS → Ask server what methods are allowed           │    │
│  │  TRACE   → Diagnostic loop-back (rarely used)            │    │
│  │  CONNECT → Tunnel connection (proxy use)                 │    │
│  └──────────────────────────────────────────────────────────┘    │
│                                                                  │
└──────────────────────────────────────────────────────────────────┘

Rules:
  -  Use NOUNS (not verbs):    /users     NOT /getUsers
  -  Use PLURAL:               /users     NOT /user
  -  Use HIERARCHY:            /users/42/orders
  -  Use HTTP methods for actions (GET, POST, PUT, DELETE)

  - eg: @app.get("/users") → Get all users
  - eg: @app.post("/users") → Create a new user
  - eg: @app.put("/users/42") → Update user with ID 42
  - eg: @app.delete("/users/42") → Delete user with ID 42

"""
```

### 3. HTTP Status Codes You Will Use Most in ML APIs

- HTTP status codes are 3-digit numbers returned by a wed sever (like FastAPI) to indicate the result of a client's request (like from a browse or API consumer).

- we see the architecture of client and sever: 
    - Client ---- request -----> Sever
    -        <---- response -----

    - where,
        - For every response, the server Provide a three digit status code.
        - This three digit status code tell the client, what are happened for client request.

        - They help to client (browser, frontend, mobile app etc) understand:
            - i).   Whether the request was successful.
            - ii).  whether something went wrong.
            - iii). and what kind of issues occurred (if any)

```python
""" 
Status Codes You Will Use Most in ML APIs:
==========================================

200 OK  --> Standard Success [ A GET or POST succeeded]
└── GET /health → {"status": "healthy"}
└── POST /predict → {"species": "setosa", "confidence": 0.97}

201 Created --> Resource Created [ After a POST that Creates something ]
└── POST /models/upload → Model uploaded successfully

204 Not Content --> Success, but no data returned [After a Delete request ]

400 Bad Request --> invalid request, Missing field, wrong data type.
└── POST /predict with invalid JSON syntax

401 Unauthorized -> No/invalid authentication |  Login required

403 Forbidden --> Authenticated but no permission | Logged in but not allowed

404 Not Found --> Resource doesn't exist | patient ID not in DB
└── GET /models/nonexistent_model

422 Unprocessable Entity  ← VERY COMMON in FastAPI
└── POST /predict with {"sepal_length": "not_a_number"}
└── FastAPI automatically returns 422 when validation fails

500 Internal Server Error
└── Model crashes during inference

502 Bad Gateway 

503 Service Unavailable (server is down or overloaded)
└── Model is still loading / not ready

"""
```

### 4. Summary

```python
""" 
┌──────────────────────────────────────────────────────────────────┐
│                  MODULE 03 — KEY TAKEAWAYS                       │
│                                                                  │
│  -   GET     → Read data (safe, idempotent, no body)             │
│  -   POST    → Create/Action (unsafe, not idempotent, has body)  │
│  -   PUT     → Full replace (unsafe, idempotent, full body)      │
│  -   PATCH   → Partial update (unsafe, partial body)             │
│  -   DELETE  → Remove (unsafe, idempotent, usually no body)      │
│                                                                  │
│  -   Safe     = Doesn't change server state                      │
│  -   Idempotent = Same result no matter how many times called    │
│                                                                  │
│  - POST is the MOST used method in ML APIs (predictions!)        │
│    - PATCH > PUT for most update scenarios (send less data)      │
│    - Status codes: 200 OK, 201 Created, 204 No Content,          │
│                   404 Not Found, 409 Conflict, 422 Validation    │
│                                                                  │
│  - Pydantic schemas differ per method:                           │
│     POST  → ModelCreate (required fields for creation)           │
│     PUT   → ModelFullUpdate (ALL fields required)                │
│     PATCH → ModelPartialUpdate (ALL fields Optional)             │
│                                                                  │
│  -   Route order: specific paths BEFORE parameterized paths      │
│  -   Use tags in FastAPI to organize docs by HTTP method         │
│                                                                  │
└──────────────────────────────────────────────────────────────────┘

"""
```
