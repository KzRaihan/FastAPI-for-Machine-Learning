## Topic: Introduction of APIs

### 1. API
- Definition of APIs
    - An API is a mechanism that allows two software applications to communicate with each other

    - API stands for Application Programming interface.

    - examples
        - customer - user
        - waiter   - api
        - kitchen  - server
        - food     - respone

    - An API acts as a messenger between a client and a server.


- Advantage of api
    - Reusability
    - Faster development
    - Easy integration
    - Secure communication
    - platform independance
    

### 2. HTTP Methods
- HTTP methods define what action the client wants to perform on the server.

- The four most important methods
    - 1. GET    -> Retrieve data
    - 2. Post   -> Create data (eg. logging facebook, searing youtube)
    - 3. Put    -> Update data
    - 4. Delete -> Delete data


- Example:
    - GET/students
    - http://127.0.0.1:8000/student (here student is a route)

### 3. JSON

### 4. Postman
- Postman is a tool used to test APIs without writing frontend code.

- It allows developers to:
    - Send API requests
    - View responses
    - Test endpoints
    - Debug APIs
