### 1. Introduction of API

- Definition: 
    - An API is a set of rules and protocols that allows one software application to communicate with another.

    - API = Application Programming Interface

    


```python
""" 
Think of it this way:

┌─────────────────────────────────────────────────────┐
│                                                     │
│   Software A  ◄────── API ──────►  Software B       │
│   (Requester)      (Contract)      (Provider)       │
│                                                     │
└─────────────────────────────────────────────────────┘

The API defines:
  • WHAT you can ask for
  • HOW you should ask for it
  • WHAT you will get back

"""
```

```python
"""   
    -> Real-World Analogy: The Restaurant
     -------------------------------------

┌──────────────────────────────────────────────────────────────┐
│                    THE RESTAURANT ANALOGY                    │
│                                                              │
│   YOU (Customer)                                             │
│     │                                                        │
│     │  1. Read the MENU (API Documentation)                  │
│     │  2. Place an ORDER (API Request)                       │
│     │                                                        │
│     ▼                                                        │
│   WAITER (API)                                               │
│     │                                                        │
│     │  3. Takes order TO the kitchen                         │
│     │  4. Brings food BACK to you                            │
│     │                                                        │
│     ▼                                                        │
│   KITCHEN (Server/Database)                                  │
│     │                                                        │
│     │  5. Prepares the food (Processes the request)          │
│     │  6. Plates it properly (Formats the response)          │
│     │                                                        │
│     ▼                                                        │
│   FOOD DELIVERED (API Response)                              │
│                                                              │
│   Key Insight: You NEVER walk into the kitchen directly.     │
│   The waiter (API) is the ONLY interface you use.            │
│                                                              │
└──────────────────────────────────────────────────────────────┘


"""

""" 
    -> Everyday API Examples we Already Use
    ------------------------------------------
┌─────────────────┬───────────────────────┬──────────────────────────┐
│  Application    │  API Used             │  What Happens            │
├─────────────────┼───────────────────────┼──────────────────────────┤
│  Google Maps    │  Maps API             │  Get directions, places  │
│  Weather App    │  Weather API          │  Get temperature, forecast│
│  Login with     │  Google/Facebook      │  Authenticate user       │
│  Google         │  OAuth API            │                          │
│  Payment        │  Stripe/PayPal API    │  Process transactions    │
│  ChatGPT        │  OpenAI API           │  Send prompt, get reply  │
└─────────────────┴───────────────────────┴──────────────────────────┘


"""
```

```python
""" -> Why APIs Are Required
============================== 
WITHOUT APIs:
-------------
┌──────────┐          ┌──────────┐
│ Mobile   │          │          │
│ App      │───???────│ Database │   Every app needs to know
├──────────┤          │          │   the database structure,
│ Web      │───???────│          │   credentials, and internal
│ App      │          │          │   logic. This is:
├──────────┤          │          │
│ Partner  │───???────│          │      -  Insecure
│ App      │          │          │      -  Tightly coupled
└──────────┘          └──────────┘      -  Hard to maintain
                                        -  Impossible to scale



WITH APIs:
----------
┌──────────┐                        ┌──────────┐
│ Mobile   │──────┐                 │          │
│ App      │      │   ┌─────────┐   │          │
├──────────┤      ├──►│         │──►│ Database │
│ Web      │──────┤   │   API   │   │          │
│ App      │      ├──►│ (Gate-  │──►│          │
├──────────┤      │   │  keeper)│   │          │
│ Partner  │──────┘   └─────────┘   └──────────┘
│ App      │
└──────────┘           -  Secure
                       -  Loosely coupled
                       -  Easy to maintain
                       -  Scalable
"""
```

```python
""" 
Five Core Reasons APIs Exist:
============================

1. ABSTRACTION
   └── Hide complex internal logic behind simple interfaces
   └── Example: You call "get_weather(city)" — you don't care HOW it works

2. SECURITY
   └── Control WHO can access WHAT data
   └── Example: API keys, authentication tokens, rate limiting

3. STANDARDIZATION
   └── Provide a consistent way for different systems to communicate
   └── Example: Any language (Python, JavaScript, Java) can call the same API

4. SCALABILITY
   └── Frontend and backend can scale independently
   └── Example: Add 10 new mobile apps without changing the server

5. REUSABILITY
   └── Build once, use everywhere
   └── Example: One payment API serves web, mobile, and partner apps

"""
```

### 2. Client-Server Architecture
- It is a model where two separate systems communicate over a network.

```python
""" 
┌─────────────────────────────────────────────────────────────────┐
│                   CLIENT-SERVER MODEL                           │
│                                                                 │
│   ┌────────────┐           Network             ┌────────────┐   │
│   │            │  ──── REQUEST ──────────────► │            │   │
│   │   CLIENT   │                               │   SERVER   │   │
│   │            │  ◄─── RESPONSE ─────────────  │            │   │
│   └────────────┘          (HTTP)               └────────────┘   │
│                                                                 │
│   Client:                          Server:                      │
│   • Initiates requests             • Listens for requests       │
│   • Displays data to user          • Processes requests         │
│   • Browser, Mobile App,           • Accesses database          │
│     Python script                  • Returns responses          │
│                                                                 │
└─────────────────────────────────────────────────────────────────┘


"""
```

### 3. REST API
- Definition:
    - REST is an architectural style (a set of guidelines) for designing APIs. It is NOT a protocol, NOT a library, NOT a framework. It is a set of constraints.
    
    - REST = Representational State Transfer


```python
""" -> The 6 Constraints of REST
================================
┌──────────────────────────────────────────────────────────────┐
│                  6 REST CONSTRAINTS                          │
│                                                              │
│  1. CLIENT-SERVER                                            │
│     └── Client and server are separate and independent       │
│                                                              │
│  2. STATELESS                                                │
│     └── Each request contains ALL information needed         │
│     └── Server does NOT remember previous requests           │
│     └── No sessions stored on server                         │
│                                                              │
│  3. CACHEABLE                                                │
│     └── Responses can be cached to improve performance       │
│                                                              │
│  4. UNIFORM INTERFACE                                        │
│     └── Consistent URL patterns and methods                  │
│     └── /users, /users/1, /users/1/orders                    │
│                                                              │
│  5. LAYERED SYSTEM                                           │
│     └── Client doesn't know if it's talking to the           │
│         actual server or a load balancer or proxy            │
│                                                              │
│  6. CODE ON DEMAND (Optional)                                │
│     └── Server can send executable code to client            │
│                                                              │
└──────────────────────────────────────────────────────────────┘


"""
```

```python
"""     -> REST Resource Design:
        ==========================

- RESOURCES are the nouns in your API:

┌─────────────────────────────────────────────────────┐
│  Resource       │  URL (Endpoint)                   │
├─────────────────┼───────────────────────────────────┤
│  All users      │  GET    /users                    │
│  Single user    │  GET    /users/42                 │
│  Create user    │  POST   /users                    │
│  Update user    │  PUT    /users/42                 │
│  Delete user    │  DELETE /users/42                 │
│  User's orders  │  GET    /users/42/orders          │
└─────────────────┴───────────────────────────────────┘

Rules:
  -  Use NOUNS (not verbs):    /users     NOT /getUsers
  -  Use PLURAL:               /users     NOT /user
  -  Use HIERARCHY:            /users/42/orders
  -  Use HTTP methods for actions (GET, POST, PUT, DELETE)

"""
```

### 4. Request and Response

```python
"""  -> Anatomy of an HTTP Request
    ==============================

┌──────────────────────────────────────────────────────┐
│                   HTTP REQUEST                       │
│                                                      │
│  ┌─────────────────────────────────────────────┐     │
│  │ REQUEST LINE                                │     │
│  │ POST /api/predict HTTP/1.1                  │     │
│  │  │      │            │                      │     │
│  │  │      │            └── HTTP Version       │     │
│  │  │      └─────────────── URL/Endpoint       │     │
│  │  └────────────────────── HTTP Method        │     │
│  └─────────────────────────────────────────────┘     │
│                                                      │
│  ┌─────────────────────────────────────────────┐     │
│  │ HEADERS (Metadata about the request)        │     │
│  │ Host: api.example.com                       │     │
│  │ Content-Type: application/json              │     │
│  │ Authorization: Bearer token123              │     │
│  │ Accept: application/json                    │     │
│  └─────────────────────────────────────────────┘     │
│                                                      │
│  ┌─────────────────────────────────────────────┐     │
│  │ BODY (Data sent with request)               │     │
│  │ {                                           │     │
│  │   "sepal_length": 5.1,                      │     │
│  │   "sepal_width": 3.5,                       │     │
│  │   "petal_length": 1.4,                      │     │
│  │   "petal_width": 0.2                        │     │
│  │ }                                           │     │
│  └─────────────────────────────────────────────┘     │
│                                                      │
└──────────────────────────────────────────────────────┘



"""
# ======================***===============================

"""   -> Anatomy of an HTTP Response:
    =================================

┌──────────────────────────────────────────────────────┐
│                   HTTP RESPONSE                      │
│                                                      │
│  ┌─────────────────────────────────────────────┐     │
│  │ STATUS LINE                                 │     │
│  │ HTTP/1.1 200 OK                             │     │
│  │    │       │   │                            │     │
│  │    │       │   └── Reason Phrase            │     │
│  │    │       └────── Status Code              │     │
│  │    └────────────── HTTP Version             │     │
│  └─────────────────────────────────────────────┘     │
│                                                      │
│  ┌─────────────────────────────────────────────┐     │
│  │ HEADERS                                     │     │
│  │ Content-Type: application/json              │     │
│  │ Content-Length: 142                         │     │
│  │ Date: Mon, 15 Jan 2025 10:30:00 GMT         │     │
│  └─────────────────────────────────────────────┘     │
│                                                      │
│  ┌─────────────────────────────────────────────┐     │
│  │ BODY                                        │     │
│  │ {                                           │     │
│  │   "prediction": 0,                          │     │
│  │   "species": "setosa",                      │     │
│  │   "confidence": 0.97                        │     │
│  │ }                                           │     │
│  └─────────────────────────────────────────────┘     │
│                                                      │
└──────────────────────────────────────────────────────┘



"""
```

```python
"""  -> Complete Request-Response Cycle
    ======================================

CLIENT                                              SERVER
  │                                                    │
  │  1. Build Request                                  │
  │  ─────────────────────────────────────────────►    │
  │  POST /api/predict                                 │
  │  Content-Type: application/json                    │
  │  {"sepal_length": 5.1, ...}                        │
  │                                                    │
  │                              2. Receive & Parse    │
  │                              3. Validate Input     │
  │                              4. Process (ML Model) │
  │                              5. Build Response     │
  │                                                    │
  │  6. Receive Response                               │
  │  ◄─────────────────────────────────────────────    │
  │  HTTP/1.1 200 OK                                   │
  │  {"species": "setosa", "confidence": 0.97}         │
  │                                                    │
  │  7. Parse & Display                                │
  │                                                    │


"""
```

### 5. JSON
- JSON = JavaScript Object Notation

- JSON is the standard data format for API communication. It is language-independent — every programming language can read and write JSON.


```python
""" -> JSON Syntax Rules:
    ========================

{   "Key" : value
    "string_value": "Hello World",
    "integer_value": 42,
    "float_value": 3.14,
    "boolean_value": true,
    "null_value": null,
    "array_value": [1, 2, 3, 4, 5],
    "nested_object": {
        "name": "FastAPI",
        "version": "1.0"
    }
}


┌─────────────────────────────────────────────────────┐
│                  JSON RULES                         │
│- Key contains:                                      │
│   - Keys MUST be strings in double quotes           │
│   - Strings use double quotes (NOT single quotes)   │
│   - Numbers: integer or floating point              │
│   - Boolean: true or false (lowercase)              │
│   - null (lowercase, NOT None)                      │
│   - Arrays: ordered list in [ ]                     │
│   - Objects: key-value pairs in { }                 │
│   - No trailing commas                              │
│   - No comments                                     │
│   - No single quotes                                │
│   - No undefined                                    │
└─────────────────────────────────────────────────────┘
"""
```

```python
# Python ↔ JSON Mapping
import json

# Python dictionary → JSON string
python_dict = {
    "name": "Iris Model",
    "accuracy": 0.95,
    "classes": ["setosa", "versicolor", "virginica"],
    "is_trained": True,
    "metadata": None
}

json_string = json.dumps(python_dict, indent=2)
print(json_string)
```

```python
# JSON string → Python dictionary
parsed_data = json.loads(json_string)
print(parsed_data["name"])  # "Iris Model"
```

### 6. HTTP

- HTTP = HyperText Transfer Protocol

- HTTP is the protocol (set of rules) that governs how messages are formatted and transmitted between client and server on the web.

```python
""" 
┌──────────────────────────────────────────────────────────────┐
│                    HTTP OVERVIEW                             │
│                                                              │
│  HTTP defines:                                               │
│    • How a client ASKS for data (Request)                    │
│    • How a server SENDS back data (Response)                 │
│    • What METHODS are available (GET, POST, PUT, DELETE)     │
│    • What STATUS CODES mean (200, 404, 500)                  │
│    • How HEADERS provide metadata                            │
│                                                              │
│  HTTP is:                                                    │
│    • Text-based protocol                                     │
│    • Stateless (no memory between requests)                  │
│    • Request-Response model                                  │
│    • Runs on port 80 (HTTP) or 443 (HTTPS)                   │
│                                                              │
└──────────────────────────────────────────────────────────────┘

"""

"""  -> HTTP Methods (Verbs) — The Big Four
    ==========================================
    
┌──────────┬───────────────┬──────────────┬─────────────────────────────┐
│  Method  │  Purpose      │  Has Body?   │  Example                    │
├──────────┼───────────────┼──────────────┼─────────────────────────────┤
│  GET     │  READ data    │  No          │  GET /users                 │
│          │  Retrieve     │              │  GET /users/42              │
│          │  Fetch        │              │  GET /models                │
├──────────┼───────────────┼──────────────┼─────────────────────────────┤
│  POST    │  CREATE data  │  Yes         │  POST /users                │
│          │  Submit       │              │  POST /predict              │
│          │  Send         │              │  POST /train                │
├──────────┼───────────────┼──────────────┼─────────────────────────────┤
│  PUT     │  UPDATE data  │  Yes         │  PUT /users/42              │
│          │  Replace      │              │  PUT /models/v2             │
│          │  Modify       │              │                             │
├──────────┼───────────────┼──────────────┼─────────────────────────────┤
│  DELETE  │  REMOVE data  │  Rarely      │  DELETE /users/42           │
│          │  Destroy      │              │  DELETE /models/old         │
│          │  Erase        │              │                             │
└──────────┴───────────────┴──────────────┴─────────────────────────────┘



"""
```

```python
""" 
HTTP vs HTTPS:
===============

HTTP:
  Client ──── plain text ────► Server
  -  Anyone can read the data in transit

HTTPS (HTTP + TLS/SSL):
  Client ──── encrypted ─────► Server
  -  Data is encrypted in transit
  -  Always use HTTPS in production

"""
```

### 7. Why ML Models Need APIs

```python
""" 
┌──────────────────────────────────────────────────────────────────┐
│           WHY ML MODELS NEED APIs                                │
│                                                                  │
│  1. ACCESSIBILITY                                                │
│     Without API: Only data scientists can run the model          │
│     With API:    Anyone/anything can use it via HTTP             │
│                                                                  │
│  2. LANGUAGE INDEPENDENCE                                        │
│     Model trained in Python, but API can be called from:         │
│     JavaScript, Java, Go, Rust, C#, or ANY language              │
│                                                                  │
│  3. SCALABILITY                                                  │
│     Run model on a powerful GPU server                           │
│     Serve thousands of clients simultaneously                    │
│     Scale horizontally with multiple instances                   │
│                                                                  │
│  4. SEPARATION OF CONCERNS                                       │
│     Data scientists → focus on MODEL quality                     │
│     Backend devs   → focus on API reliability                    │
│     Frontend devs  → focus on USER experience                    │
│                                                                  │
│  5. VERSIONING                                                   │
│     /api/v1/predict → Old model                                  │
│     /api/v2/predict → New improved model                         │
│     Both can run simultaneously (A/B testing)                    │
│                                                                  │
│  6. MONITORING & LOGGING                                         │
│     Track every prediction request                               │
│     Monitor latency, errors, input distributions                 │
│     Detect model drift over time                                 │
│                                                                  │
│  7. SECURITY                                                     │
│     API keys and authentication                                  │
│     Rate limiting (prevent abuse)                                │
│     Input validation (prevent malicious inputs)                  │
│                                                                  │
└──────────────────────────────────────────────────────────────────┘

"""


""" -> Real-World ML API Examples
=================================

┌───────────────────┬──────────────────────────────────────────┐
│  Company/Service  │  What Their ML API Does                  │
├───────────────────┼──────────────────────────────────────────┤
│  OpenAI           │  Text generation, embeddings, vision     │
│  Google Vision    │  Image classification, OCR               │
│  AWS Rekognition  │  Face detection, object recognition      │
│  Hugging Face     │  NLP inference (sentiment, translation)  │
│  Stripe Radar     │  Fraud detection on payments             │
│  Spotify          │  Music recommendation                    │
│  Netflix          │  Content recommendation                  │
│  Tesla            │  Autonomous driving decisions            │
└───────────────────┴──────────────────────────────────────────┘

All of these serve ML models through APIs!


"""
```

### 8. FastAPI for Machine Learning

```python
"""
┌──────────────────────────────────────────────────────────────┐
│                     FASTAPI                                  │
│                                                              │
│  FastAPI is a modern, fast, web framework for building       │
│  APIs with Python, based on standard Python type hints.      │
│                                                              │
│  Created by: Sebastián Ramírez (@tiangolo)                   │
│  First Release: December 2018                                │
│  Language: Python 3.7+                                       │
│  Type: ASGI Framework                                        │
│                                                              │
└──────────────────────────────────────────────────────────────┘

"""

""" 
-> Why FastAPI Over Other Frameworks?
    =================================

┌──────────────────┬───────────┬───────────┬───────────────────┐
│  Feature         │  Flask    │  Django   │  FastAPI          │
├──────────────────┼───────────┼───────────┼───────────────────┤
│  Speed           │  Slow     │  Medium   │  Very Fast        │
│  Async Support   │  Limited  │  Limited  │  Native           │
│  Type Hints      │  No       │  No       │  Core Feature     │
│  Auto Docs       │  No       │  No       │  Built-in         │
│  Validation      │  Manual   │  Forms    │  Automatic        │
│  Learning Curve  │  Easy     │  Steep    │  Easy             │
│  ML Friendly     │  OK       │  Complex  │  Excellent        │
│  Performance     │  ~400rps  │  ~800rps  │  ~5000rps         │
│  Modern Python   │  No       │  Partial  │  Yes              │
└──────────────────┴───────────┴───────────┴───────────────────┘

rps = requests per second (approximate, depends on use case)

"""
```

```python
""" -> The 5 Pillars of FastAPI:

┌──────────────────────────────────────────────────────────────┐
│                                                              │
│  1. FAST PERFORMANCE                                         │
│     Built on Starlette (ASGI) and Uvicorn                    │
│     One of the fastest Python frameworks                     │
│     On par with NodeJS and Go                                │
│                                                              │
│  2. FAST TO CODE                                             │
│     Reduce code duplication                                  │
│     Intuitive design                                         │
│     Fewer bugs                                               │
│                                                              │
│  3. AUTOMATIC DOCUMENTATION                                  │
│     Swagger UI at /docs                                      │
│     ReDoc at /redoc                                          │
│     OpenAPI schema auto-generated                            │
│                                                              │
│  4. TYPE SAFETY                                              │
│     Python type hints drive everything                       │
│     Editor support (autocomplete, error checking)            │
│     Runtime validation via Pydantic                          │
│                                                              │
│  5. STANDARDS-BASED                                          │
│     OpenAPI (formerly Swagger) specification                 │
│     JSON Schema                                              │
│     OAuth2 support                                           │
│                                                              │
└──────────────────────────────────────────────────────────────┘


"""
```

```python
""" 
FastAPI Architecture:
====================
                    FastAPI Stack
┌─────────────────────────────────────────┐
│              YOUR CODE                  │
│         (Endpoints, Logic)              │
├─────────────────────────────────────────┤
│             FastAPI                     │
│    (Routing, Validation, Docs)          │
├─────────────────────────────────────────┤
│             Pydantic                    │
│      (Data Validation & Parsing)        │
├─────────────────────────────────────────┤
│             Starlette                   │
│     (ASGI Framework, Middleware)        │
├─────────────────────────────────────────┤
│             Uvicorn                     │
│        (ASGI Server, Runs App)          │
├─────────────────────────────────────────┤
│           HTTP / Network                │
└─────────────────────────────────────────┘

"""
```

### 9. Summary

```python
""" 
┌──────────────────────────────────────────────────────────────────┐
│                  MODULE 01 — KEY TAKEAWAYS                       │
│                                                                  │
│  -> API = Contract between software systems                      │
│  -> APIs provide abstraction, security, scalability              │
│  -> Client-Server = Two independent systems communicating        │
│  -> REST = Architectural style with 6 constraints                │
│  -> HTTP = Protocol for web communication                        │
│  -> HTTP Methods: GET (read), POST (create),                     │
│                   PUT (update), DELETE (remove)                  │
│  -> Status Codes: 2xx (success), 4xx (client error),             │
│                   5xx (server error)                             │
│  -> JSON = Standard data format for APIs                         │
│  -> Endpoints = HTTP Method + URL Path                           │
│  -> ML models NEED APIs for accessibility and scalability        │
│  -> FastAPI = Modern, fast, Python-native API framework          │
│  -> FastAPI is ideal for ML because of Python, Pydantic,         │
│     auto-docs, async support, and production readiness           │
│                                                                  │
└──────────────────────────────────────────────────────────────────┘

"""
```
